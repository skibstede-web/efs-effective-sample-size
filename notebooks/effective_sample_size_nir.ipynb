## 1. Scientific Background

Effective sample size in diffuse reflectance NIR is the amount of material that meaningfully contributes to the measured spectral signal, not merely the geometric volume beneath a probe. In particulate media, photons undergo multiple scattering and wavelength-dependent absorption; therefore, signal contribution depends on optical transport behavior as well as local material composition.

The geometric probe footprint alone is insufficient because the interaction depth is constrained by optical penetration and by the scattering regime of the powder. Thus, the sampled region is an optical interaction zone, not a simple cylindrical cutout.

Powder motion further changes sampling meaning. For static beds, sampled mass is tied to an interaction zone at an instant. For flowing or moving beds, additional mass can pass through that zone during acquisition time, so the cumulative contributing mass may exceed the instantaneous in-zone mass.

A key distinction for later modeling is: (i) mass present in the optical interaction zone at one instant, versus (ii) mass that contributes over the full detector integration/acquisition window.

## 4. Conceptual Definitions

### Rigorous definitions (model-facing)
- **Probe footprint / probe window diameter**: Geometric aperture at the powder interface that bounds first-order illuminated area.
- **Optical penetration depth**: Effective depth scale over which photons significantly contribute to detected diffuse reflectance; wavelength- and matrix-dependent.
- **Optical interaction volume**: Spatial region with non-negligible contribution weighting to the measured signal.
- **Sampled mass**: Mass physically present within a defined interaction volume at a specified time.
- **Effective sample mass**: Signal-weighted equivalent mass contributing to the measurement after accounting for nonuniform photon path weighting and/or dynamics.
- **Effective number of particles**: Equivalent statistically independent particle count represented by the measured signal.
- **Acquisition time**: Detector integration or averaging interval over which signal is accumulated.
- **Powder residence time in the measurement zone**: Characteristic time a powder element remains inside the interaction zone.
- **Particle re-sampling**: Repeated contribution of the same particle(s) within an acquisition window due to local recirculation, slow movement, or repeated exposure.
- **Representativity**: Degree to which measured material contributions reflect the target bulk composition distribution.
- **Sampling error**: Variability induced by finite and potentially biased particulate sampling.

### Practical engineering definitions (implementation-facing)
- Use measurable surrogate parameters (e.g., probe diameter, penetration-depth estimate, bulk density, flow velocity, residence-time estimate) to approximate the above rigorous constructs.
- Track where practical definitions are approximations and where they are mechanistic closures.

## 6. Core Equations Framework (Placeholders)

### 6.1 Cross-sectional area (circular geometry)
$$A = \frac{\pi D^2}{4}$$
- Variables: $A$ area, $D$ diameter
- Units: $A$ in cm$^2$ (or m$^2$), $D$ in cm (or m)
- Assumption note: Circular cross section
- Classification: Mechanistic

### 6.2 Superficial velocity from mass flow
$$u_s = \frac{\dot{m}}{\rho_b A}$$
- Variables: $u_s$ superficial velocity, $\dot{m}$ mass flow rate, $\rho_b$ bulk density, $A$ flow area
- Units: $u_s$ in cm/s (or m/s)
- Assumption note: Steady flow, uniform bulk density over area
- Classification: Mechanistic (continuity-based)

### 6.3 Static sampled volume approximation
$$V_{\mathrm{static}} \approx A_{\mathrm{probe}}\,\delta_{\mathrm{opt}}$$
- Variables: $V_{\mathrm{static}}$, probe footprint area $A_{\mathrm{probe}}$, optical penetration depth $\delta_{\mathrm{opt}}$
- Units: volume in cm$^3$
- Assumption note: Uniform interaction depth over footprint
- Classification: Approximate

### 6.4 Static sampled mass approximation
$$m_{\mathrm{static}} \approx \rho_b\,V_{\mathrm{static}}$$
- Variables: $m_{\mathrm{static}}$, $\rho_b$, $V_{\mathrm{static}}$
- Units: mass in g (or kg)
- Assumption note: Homogeneous bulk density in interaction region
- Classification: Approximate

### 6.5 Dynamic sampled volume during acquisition
$$V_{\mathrm{dyn}} \approx Q_v\,t_{\mathrm{acq}}$$
with volumetric flow estimate
$$Q_v = \frac{\dot{m}}{\rho_b}$$
- Variables: $V_{\mathrm{dyn}}$, $Q_v$, acquisition time $t_{\mathrm{acq}}$
- Units: $V_{\mathrm{dyn}}$ in cm$^3$, $t_{\mathrm{acq}}$ in s
- Assumption note: Throughput-linked contribution; requires geometry-specific correction later
- Classification: Semi-mechanistic placeholder

### 6.6 Dynamic sampled mass
$$m_{\mathrm{dyn}} \approx \rho_b\,V_{\mathrm{dyn}}$$
- Variables: $m_{\mathrm{dyn}}$, $\rho_b$, $V_{\mathrm{dyn}}$
- Units: g or kg
- Assumption note: Effective contribution mass, not necessarily unique mass without re-sampling correction
- Classification: Approximate

### 6.7 Particle count estimate
$$N_{\mathrm{eff}} \sim \frac{m_{\mathrm{eff}}}{m_p}$$
- Variables: $N_{\mathrm{eff}}$ effective number of particles, $m_{\mathrm{eff}}$ effective mass, $m_p$ mean particle mass
- Units: count (dimensionless)
- Assumption note: Equivalent-particle approximation
- Classification: Approximate

### 6.8 Sampling variance scaling concept
$$\sigma_{\mathrm{sampling}}^2 \propto \frac{1}{N_{\mathrm{eff}}}$$
- Variables: sampling variance proxy, effective particle count
- Units: model-dependent
- Assumption note: Qualitative scaling; constants and heterogeneity terms to be introduced later
- Classification: Conceptual/statistical

# Parameter Registry and Unit System

This section defines the structural framework for parameter handling before any scientific calculations are implemented. The project uses a structured parameter registry to maintain traceability between physical inputs, modeling assumptions, and quantities that will be derived later.

User-facing engineering units may differ from internal computational units. To support consistency and reproducibility, all internal calculations will later be performed in SI units.

Inputs, assumptions, and derived quantities must remain clearly separated so model logic can be audited, validated, and extended without ambiguity.

## Design Principles for the Parameter Registry

A layered parameter registry is preferred over a flat dictionary because it separates scientific concerns and makes model intent explicit. This improves maintainability as the notebook evolves from conceptual scaffolding into deterministic and then scenario-based workflows.

SI units are preferred internally because they provide a single canonical basis for cross-setup comparison and reduce hidden conversion errors. Assumptions must be explicit so interpretation remains tied to the model version and its simplifications.

Derived quantities should not be mixed with user inputs. Inputs represent controllable or measured values, while derived quantities are model outputs that depend on equations and assumptions. Keeping these separate prevents circular dependencies and accidental overwrites.

Setup-specific sub-registries are important because each interface introduces distinct geometry and kinematic parameters. Parameter names should include units to make dimensional intent visible at the registry-key level.

The registry is designed to support four setups: Static, Spoon, Spiderwheel, and Pipe.

## Recommended Naming Convention

Parameter keys should follow:

`parameter_name_unit`

Examples:

- `probe_diameter_m`
- `penetration_depth_m`
- `bulk_density_kg_m3`
- `mass_flow_rate_kg_s`
- `acquisition_time_s`

Short symbolic names such as `d`, `rho`, or `m_dot` should not be used as registry keys because they are ambiguous outside equation context and reduce readability in collaborative scientific workflows.
